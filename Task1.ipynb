```python
import pandas as pd

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)

print(df.head())
print(df.shape)
df.to_csv("Titanic.csv", index=False)
```

```python
df.info()
     
```

```python
df.isnull().sum()
```

```python
"""
    Observation

The dataset contains missing values in the following columns:

    Age: 177 missing values
    Cabin: 687 missing values
    Embarked: 2 missing values

The Cabin column has a very large number of missing values, while Age and Embarked have comparatively fewer missing values. 
These missing values will be handled in the next step.
"""
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())

df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

df.drop("Cabin", axis=1, inplace=True)
```

```python
"""
Observation

The missing values were successfully handled.

    Age values were filled using the median.
    Embarked values were filled using the mode.
    Cabin column was removed from the dataset.
"""
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x="Survived", data=df)
plt.title("Survival Count")
plt.savefig("survival_count.png")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x="Sex", hue="Survived", data=df)
plt.title("Survival by Gender")
plt.savefig("survival_by_gender.png")
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.histplot(df["Age"], bins=20)
plt.title("Age Distribution")
plt.savefig("age_distribution.png")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x="Pclass", data=df)
plt.title("Passenger Class Distribution")
plt.savefig("passsenger_class_distribution.png")
plt.show()
```
